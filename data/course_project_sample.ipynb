# 0️⃣ 서론

코로나19에 의한 경기침체, 부동산 3법 등 강력한 부동산 규제 정책에도 주택가격은 상승했다. 어려운 상황에서도 상승하는 주택 가격에 부동산 투자에 대한 관심이 쏠리고, 국토교통부에 따르면 주택 거래량은 증가하여 지난해 전국에서 거래된 토지거래량은 토지 필지 수 기준 350만7000개로 1997년 통계 집계 이래 최대를 기록했다.[[참고]](http://news.kmib.co.kr/article/view.asp?arcid=0015468650&code=61141511&cp=nv)

이렇게 많은 부동산이 거래되고 있다는 사실을 접하고 대체 어디서 얼마나 많은 부동산들이 어떤 가격에 거래되고 있는지 궁금증이 생겼다. 우리팀은 특히나 집값이 높은 편인 서울특별시를 중심으로 부동산 거래 현황을 파악해보고자 했다. 데이터를 통해 작년에 최대를 기록한 거래량의 과거부터 최근까지 추이를 직접 확인하고 주택 가격은 어떤 변화를 보였는지 또한 어떤 부동산들이 어떤 거래유형으로 거래되었는지 등을 파악한 후, 거래건들의 다양한 정보를 활용하여 특정 조건을 만족하는 주택을 찾아보는 작업까지 해볼 것이다.

본격적인 분석에 들어가기에 앞서 분석에 사용되는 부동산 용어에 대해 정리하도록 한다.

* 거래유형 : 매매, 전세, 월세, 준전세, 준월세

> 매매 - 부동산의 매도인과 매수인이 그 소유권의 변동을 목적으로 하는 계약을 체결ㆍ이행하고 소유권이전을 행하는 것을 말한다.

> 전세 - 집주인(임대인)과 세입자 간 임대차 계약을 통해 집을 일정 기간 빌리는 것으로, 주택가격의 60~80%에 해당하는 금액을 전세금으로 하여 집주인(임대인)에게 예탁하고 계약 기간이 끝나면 전세금을 100% 돌려받고 나가는 것을 말한다.

> 월세 - 집주인(임대인)과 세입자 간 임대차 계약을 통해 집을 일정 기간 빌리는 것으로, 세입자는 일정액의 보증금을 집주인에게 예탁하고 매달 일정 날짜에 집주인한테 월 차임을 납부하며, 계약 기간이 끝나면 보증금만 돌려받고 나가는 것을 말한다. 월세는 보증금이 월세의 12배 이하이다.

> 준전세 - 월세와 같은 형식으로 보증금과 월세를 각각 예탁, 납부하지만 보증금이 월세의 240배 초과하는 경우를 말한다.

> 준월세 - 월세와 같은 형식으로 보증금과 월세를 각각 예탁, 납부하지만 보증금이 월세의 12~240배인 경우를 말한다.

* 건물주용도 : 오피스텔, 아파트, 단독주택/다가구주택, 연립주택/다세대주택

>오피스텔 - 오피스(Office)와 호텔(Hotel)의 합성어로 주거 공간과 사무용도로 모두 사용할 수 있는 건물을 말한다.

> 아파트 - 여러 세대가 거주하는 형식으로 5층 이상의 공동주택을 말한다.

단독/다가구<br>
> 단독주택 - 한 세대가 하나의 건축물 안에서 독립된 주거생활을 할 수 있는 구조로 된 주택을 말한다.

> 다가구주택 -주택으로 쓰이는 층수가 3개 층 이하이고, 1개 동의 주택으로 쓰는 바닥면적(지하주차장 면적 제외)의 합계가 660m2 이하이며, 19세대 이하가 거주할 수 있는 주택을 말한다. 다가구주택은 「건축법」에 의한 용도별 건축물의 종류상 단독주택에 해당한다.

연립/다세대<br>
> 연립주택 - 주택으로 쓰이는 1개 동의 바닥면적(2개 이상의 동을 지하주차장으로 연결하는 경우에는 각각의 동으로 본다.)의 합계가 660㎡ 초과이고, 층수가 4개 층 이하인 주택을 말한다.

> 다세대주택 - 주택으로 쓰이는 1개 동의 바닥면적(2개 이상의 동을 지하주차장으로 연결하는 경우에는 각각의 동으로 본다.)의 합계가 660㎡ 이하이고, 층수가 4개 층 이하인 주택을 말한다.

## 1-1) 데이터 불러오기

```python
import pandas as pd
from urllib.parse import quote

# 데이터는 GitHub 저장소에서 직접 읽어온다 (로컬에 파일이 없어도 동작)
GITHUB_BASE = 'https://raw.githubusercontent.com/JeongYunLee/data-2026-1/main/data/'
def data_url(fname):
    # 한글 파일명을 URL 인코딩
    return GITHUB_BASE + quote(fname)

# 사용 컬럼만 추려 gzip 압축한 CSV (pandas가 .gz를 자동 인식)
df1 = pd.read_csv(data_url('실거래가_2018-2020.csv.gz'), encoding='utf-8-sig')
df2 = pd.read_csv(data_url('전월세_2018-2020.csv.gz'), encoding='utf-8-sig')
```

## 2-1) 거래유형 별 가격 현황

매매, 전세(준전세), 월세(준월세) 가격이 최근 3년동안 어떻게 변화하였는가?

1. 각 연도별 매매가 현황
2. 각 연도별 전세 현황
3. 각 연도별 준전세 현황
4. 각 연도별 월세 현황
5. 각 연도별 준월세 현황
   
   
* swarmplot 사용
* 월세, 준월세, 준전세는 '보증금+한달 치 월세' 합

```python
## 매매 연도별 가격
pd.options.display.float_format = '{:.5f}'.format  # 지수표현 없애기
buy_1 = df1[['신고년도', '물건금액']].rename(columns={'신고년도': '연도', '물건금액': '가격'}).copy()
buy_1['가격'] = buy_1['가격'] / 10000  # 단위 만원
buy_1['거래유형'] = '매매'
buy_1.head()
```

```python
# 매매 연도별 거래량
df1.groupby('신고년도')['ID'].count()
```

2018, 2019년에 비해서 큰 차이는 아니지만, 2020년에는 매매 거래량이 많았다.

```python
## 전세·준전세·월세·준월세 연도별 가격 (보증금 + 임대료)
monthly_yearly_1 = pd.DataFrame({
    '거래유형': df2['전월세구분'],
    '연도': df2['접수년도'].astype(int),
    '가격': df2['보증금'] + df2['임대료'],
})
monthly_yearly_1.head()
```

```python
# 전월세 연도별 거래량
df2.groupby('접수년도')['ID'].count()
```

거래횟수를 보면 2020년은 2018, 2019년도에 비해 전월세 거래 횟수가 현저히 적다. 앞서 매매 거래량과 정반대의 현상이다.

```python
# 거래유형·연도별 거래량
df2.groupby(['전월세구분', '접수년도'])['ID'].count()
```

매매와 합친 데이터에서도 2020년의 매매, 전월세 횟수가 2018, 2019에 비해 적게 나타나는데 이는 전월세 데이터의 영향인 것을 알 수 있다.


```python
## 매매 + 전월세 전체 합치기
eda_1 = pd.concat([buy_1, monthly_yearly_1])
eda_1.reset_index(drop=True, inplace=True)
print(eda_1.groupby(['연도']).count())
```

전체 데이터가 140만 행이 넘는데, 각각의 행을 dot으로 찍는 시각화를 시도했지만 seaborn, plotly 모두 실행이 안된다. (런타임 끊김) 따라서 만개의 데이터만 random 하게 추출하여 시각화를 진행하였다.

```python
## 전체 약 145만 행을 모두 점으로 찍으면 메모리 초과로 그려지지 않는다.
## 따라서 1만 개만 무작위 추출해 시각화한다.
eda_1 = eda_1.sample(n=10000, random_state=1)
print(eda_1.groupby(['연도']).count())
```

```python
import plotly.express as px

fig = px.strip(eda_1, x='연도', y='가격', color='연도', facet_col='거래유형')
fig.show()
```

매매의 가격이 다른 거래유형에 비해 지나치게 높아서 비교하기 어렵다.    
따라서 각 거래유형별로 가격을 정규화시켜주었다.

```python
## 정규화 (거래유형별 가격 분포 비교)
from sklearn import preprocessing

def scaled(type):
    eda_1_1 = eda_1[eda_1['거래유형'] == type]
    wow = eda_1_1[['가격']].values
    wow_scaled = preprocessing.MinMaxScaler().fit_transform(wow.astype(float))
    out = eda_1_1.copy()
    out['가격_정규화'] = wow_scaled
    return out.sort_values(by=['연도'], ascending=True)

eda_1_scaled = pd.concat([scaled('매매'), scaled('전세'), scaled('준전세'), scaled('월세'), scaled('준월세')], axis=0)
eda_1_scaled
```

```python
fig = px.strip(eda_1_scaled, x='연도', y='가격_정규화', color='연도', facet_col='거래유형')
fig.show()
```

처음에 세운 가설은 '시간이 지날수록 주택가격은 상승세를 보일것이다' 이다. 3개년을 비교했을 때, 2018년에 비해 2020년의 주택가격이 월등하게 차이가 나진 않지만, **매매의 경우에는 2018년에서 2020년으로 갈수록 가격이 높은 곳에 더 많은 분포를 보이고 있다.** 준전세, 월세, 준월세 거래는 최고 거래가가 2018년에 비해 2019, 2020년에 적은 현황을 보이고, 전세같은 경우에도 가격 분포가 시간이 지날수록 오히려 약하지만 낮은 곳에 분포하고 있다.      


그러나 최근 3개년의 데이터만 확인 했을 때는 뚜렷한 거래가격의 변화 및 추이를 확인할 순 없다.   

아래는 각각의 거래현황 별로 가격분포를 더 잘 확인해보기 위해 그린 정규화하지 않은 그래프이다.

## 2-2) 건물면적별 가격 현황

매매, 전세(준전세), 월세(준월세) 가격은 건물 면적별로 최근 3년동안 어떻게 변화하였는가?

1. 각 연도별 건물면적별(㎡) 매매가 현황
2. 각 연도별 건물면적별(㎡) 전세가 현황
3. 각 연도별 건물면적별(㎡) 준전세가 현황
4. 각 연도별 건물면적별(㎡) 월세가 현황
5. 각 연도별 건물면적별(㎡) 준월세가 현황


* scatter chart
* 각 연도별 색깔 다르게

```python
## 매매: 연도·가격·면적
buy_2 = df1[['신고년도', '물건금액', '건물면적']].rename(
    columns={'신고년도': '연도', '물건금액': '가격', '건물면적': '면적'}).copy()
buy_2['가격'] = buy_2['가격'] / 10000  # 만원
buy_2['거래유형'] = '매매'
buy_2.head()
```

```python
## 전세~준월세: 연도·가격·면적
monthly_yearly_2 = pd.DataFrame({
    '거래유형': df2['전월세구분'],
    '연도': df2['접수년도'].astype(int),
    '가격': df2['보증금'] + df2['임대료'],
    '면적': df2['임대면적'],
})
monthly_yearly_2.head()
```

```python
eda_2 = pd.concat([buy_2, monthly_yearly_2])
eda_2.reset_index(drop=True, inplace=True)
eda_2 = eda_2.sample(n=10000, random_state=1)
eda_2
```

```python
# 전월세 데이터 분리(전세,월세,준전세,준월세)
jeonse = month_jeonse[month_jeonse['전월세구분']=='전세']
month = month_jeonse[month_jeonse['전월세구분']=='월세']
quasiJeonse = month_jeonse[month_jeonse['전월세구분']=='준전세']
quasiMonth = month_jeonse[month_jeonse['전월세구분']=='준월세']
```

```python
# 건물주용도별로 데이터 분리하는 함수 정의
def house_divide(df):
    office =df[df['건물주용도']=='오피스텔']
    apt = df[df['건물주용도']=='아파트']
    tmp = df['건물주용도'].unique()
    if '단독/다가구' in tmp:
        house = df[df['건물주용도']=='단독/다가구']
        multi = df[df['건물주용도']=='다세대/연립']
    else:
        house = df[df['건물주용도']=='단독주택']
        multi = df[df['건물주용도']=='연립주택']
    return office,apt,house,multi

# 연도별 거래유형별 거래량 데이터 프레임 만드는 함수 정의
def make_count(df1,df2,df3,df4,df5):
    b = df1.groupby(by = '연도').count()['건물주용도'] #매매
    j = df2.groupby(by = '연도').count()['건물주용도'] #전세
    m = df3.groupby(by = '연도').count()['건물주용도'] #월세
    qj = df4.groupby(by = '연도').count()['건물주용도'] #준전세
    qm = df5.groupby(by = '연도').count()['건물주용도'] #준월세

    all_df = pd.concat([b,j,m,qj,qm],axis=1,keys = ['매매','전세','월세','준전세','준월세'])
    all_df.index = ['2018년','2019년','2020년']
    return all_df
```

```python
# 매매,전세,월세,준전세,준월세 각각 건물주용도별로 df 분리
buy_office, buy_apt, buy_house, buy_multi = house_divide(buy)
jeonse_office,jeonse_apt,jeonse_house,jeonse_multi = house_divide(jeonse)
month_office,month_apt,month_house,month_multi = house_divide(month)
quasiJeonse_office,quasiJeonse_apt,quasiJeonse_house,quasiJeonse_multi = house_divide(quasiJeonse)
quasiMonth_office,quasiMonth_apt,quasiMonth_house,quasiMonth_multi = house_divide(quasiMonth)
```

```python
# 각 거래량 count한 df 만들기
all_office = make_count(buy_office,jeonse_office,month_office,quasiJeonse_office,quasiMonth_office)
all_house = make_count(buy_house,jeonse_house,month_house,quasiJeonse_house,quasiMonth_house)
all_apt = make_count(buy_apt,jeonse_apt,month_apt,quasiJeonse_apt,quasiMonth_apt)
all_multi = make_count(buy_multi,jeonse_multi,month_multi,quasiJeonse_multi,quasiMonth_multi)
```

3. 2020년 거래유형 별 (매매, 월세, 준월세, 전세, 준전세) 거래량 비율 비교

2-4-1) 데이터 전처리

```python
# [연도,거래량,거래유형,건물주용도]를 컬럼으로 하여 데이터 구조를 바꾸는 함수
def change_df(dfs,string):
    output = pd.DataFrame()
    for i in range(0,5):
        obj = dfs.iloc[:,i]
        val = obj.name
        tmp = obj.reset_index()
        tmp.columns = ['연도','거래량']
        tmp['거래유형'] = val
        output = pd.concat([output,tmp])
    output.reset_index(inplace=True,drop=True)
    output.sort_values(by='연도',inplace=True)
    output['건물주용도'] = string

    df_18 = output.iloc[:5]
    df_19 = output.iloc[5:10]
    df_20 = output.iloc[10:]

    return df_18,df_19,df_20
```

```python
df_18 = pd.DataFrame()
df_19 = pd.DataFrame()
df_20 = pd.DataFrame()
for i,s in zip(items,['오피스텔','아파트','단독/다가구','연립/다세대']):
    t1,t2,t3 = change_df(i,s)
    df_18 = pd.concat([df_18,t1]).reset_index(drop=True)
    df_19 = pd.concat([df_19,t2]).reset_index(drop=True)
    df_20 = pd.concat([df_20,t3]).reset_index(drop=True)
```

```python
# 거래유형별 비율을 구하는 함수 - groupby와 함께 쓸 것
def make_percents(df):
    a = (df['거래량'] / df['거래량'].sum()) *100
    return a
```

```python
df_18['퍼센트'] = (df_18.groupby(by='거래유형').apply(make_percents).reset_index()
                    .sort_values(by='level_1').set_index('level_1').drop(['거래유형'],axis=1))
df_19['퍼센트'] = (df_19.groupby(by='거래유형').apply(make_percents).reset_index()
                    .sort_values(by='level_1').set_index('level_1').drop(['거래유형'],axis=1))
df_20['퍼센트'] = (df_20.groupby(by='거래유형').apply(make_percents).reset_index()
                    .sort_values(by='level_1').set_index('level_1').drop(['거래유형'],axis=1))
```

geojson 폴리곤에서 **자치구별 중심 좌표**를 계산하고, 고객 추천 매물을 지도에 표시할 때 쓸
위치 추정 함수(`gu_geocode`)를 정의한다.

> 실거래가/전월세 데이터에는 위·경도가 없으므로, 건물은 해당 **자치구 중심 부근**에 근사 표시한다.
> (원본은 이 부분에서 구글/네이버 지오코딩 API로 실제 좌표를 받아왔다.)

```python
import hashlib

# geojson 폴리곤 -> 자치구별 대표 좌표(중심) 계산
def _gu_centroid(geom):
    if geom['type'] == 'Polygon':
        ring = geom['coordinates'][0]
    else:  # MultiPolygon: 점이 가장 많은(가장 큰) 외곽 링 사용
        ring = max((poly[0] for poly in geom['coordinates']), key=len)
    lats = [pt[1] for pt in ring]
    lons = [pt[0] for pt in ring]
    return sum(lats) / len(lats), sum(lons) / len(lons)

GU_CENTROID = {ft['properties']['name']: _gu_centroid(ft['geometry'])
               for ft in municipalities_geo['features']}


def _jitter(seed, scale=0.008):
    """같은 문자열은 항상 같은 작은 오프셋(±scale)을 반환한다.
    마커가 한 점에 겹치지 않게 흩어주며, 해시 기반이라 실행할 때마다 위치가 동일하게 재현된다."""
    h = int(hashlib.md5(str(seed).encode('utf-8')).hexdigest(), 16)
    return (((h % 1000) / 1000 - 0.5) * 2 * scale,
            ((h // 1000 % 1000) / 1000 - 0.5) * 2 * scale)


def gu_geocode(query, gu=None):
    """자치구 중심 좌표(geojson 기반)에 작은 오프셋을 더해 (위도, 경도)를 만든다.

    gu 를 직접 주면 그 자치구를, 아니면 query 문자열에 들어 있는 자치구명을 사용한다.
    (건물 정확 위치가 아니라 '자치구 중심 부근' 근사값이다.)
    """
    q = str(query)
    if gu is None:
        gu = next((g for g in GU_CENTROID if g in q), None)
    lat, lon = GU_CENTROID.get(gu, (37.5665, 126.9780))  # 못 찾으면 서울 시청 부근
    dlat, dlon = _jitter(q)
    return lat + dlat, lon + dlon
```

```python
def make_chart(gu_3, point, b_type):
    if point in ('보증금+임대료', '물건금액'):
        point = '거래액 평균'
    max_count = gu_3[point].max()
    fig = px.choropleth_mapbox(gu_3, geojson=municipalities_geo,
                  locations='id', color=point,
                  color_continuous_scale='OrRd', range_color=(0, max_count),
                  featureidkey='id', mapbox_style='carto-positron',
                  center={'lat': 37.5581, 'lon': 126.9766}, zoom=9.3,
                  animation_frame='년도', hover_name='자치구명',
                  hover_data=[point, '년도'])
    fig.update_layout(title={'text': f'<b>자치구별 {b_type} {point} 변화</b>', 'x': 0.5})
    fig.show()
```

```python
# table1(매매) 자치구별 거래량 -> 지도
def _add_id_year(gu_2018, gu_2019, gu_2020):
    name2id = {ft['properties']['name']: ft['id'] for ft in municipalities_geo['features']}
    for d, y in [(gu_2018, '2018'), (gu_2019, '2019'), (gu_2020, '2020')]:
        d['id'] = [name2id.get(n) for n in d['자치구명']]
        d['년도'] = y
    gu_3 = pd.concat([gu_2018, gu_2019, gu_2020]).reset_index(drop=True)
    return gu_3

def quantitiy(point, b_type):
    def cnt(yr):
        g = (df1[df1['신고년도'] == yr].groupby('자치구명').size()
             .reset_index(name=point).sort_values('자치구명'))
        return g
    gu_3 = _add_id_year(cnt(2018), cnt(2019), cnt(2020))
    make_chart(gu_3, point, b_type)
```

```python
# table2(전세/월세/준전세/준월세) 자치구별 거래량 -> 지도
def t2_quantitiy(point, b_type):
    def cnt(yr):
        sub = df2[(df2['접수년도'] == yr) & (df2['전월세구분'] == b_type)]
        return (sub.groupby('자치구명').size().reset_index(name=point).sort_values('자치구명'))
    gu_3 = _add_id_year(cnt(2018), cnt(2019), cnt(2020))
    make_chart(gu_3, point, b_type)
```

```python
# table1(매매) 자치구별 거래액 평균 -> 지도
def avg_price(point, b_type):
    def agg(yr):
        sub = df1[df1['신고년도'] == yr]
        g = sub.groupby('자치구명')[point].agg(['mean', 'sum', 'count']).reset_index()
        g.columns = ['자치구명', '거래액 평균', '총 거래금액', '총 거래량']
        return g.sort_values('자치구명')
    gu_3 = _add_id_year(agg(2018), agg(2019), agg(2020))
    make_chart(gu_3, point, b_type)
```

```python
# table2 자치구별 거래액 평균(보증금+임대료) -> 지도
def t2_avg_price(point1, point2, b_type):
    tmp = df2.copy()
    tmp['_price'] = tmp[point1] + tmp[point2]
    def agg(yr):
        sub = tmp[tmp['접수년도'] == yr]
        g = sub.groupby('자치구명')['_price'].agg(['mean', 'sum', 'count']).reset_index()
        g.columns = ['자치구명', '거래액 평균', '총 거래금액', '총 거래량']
        return g.sort_values('자치구명')
    gu_3 = _add_id_year(agg(2018), agg(2019), agg(2020))
    make_chart(gu_3, '보증금+임대료', b_type)
```

```python
# 건물명 -> (자치구 중심 기반) 위경도. 원본은 네이버 지역검색 + 구글 지도 API를 사용했다.
def make_address(df):
    house_address, house_lat, house_lng = [], [], []
    for name in df['건물명']:
        la, lo = gu_geocode(str(name), gu='노원구')   # 고객2는 모두 노원구 매물
        house_address.append('서울특별시 노원구')
        house_lat.append(la)
        house_lng.append(lo)
    df = df.copy()
    df['주소'] = house_address
    df['위도'] = house_lat
    df['경도'] = house_lng
    return df
```

고객3은 연식, 지역은 상관 없고 단독주택이되, 50평 이상인 집을 원한다.
예산은 40~45억이라고 한다. 데이터베이스를 활용해 고객이 원하는 단독주택 중
면적이 넓은 순위로 20개의 집을 추린다. 단독주택은 주소를 제공하지 않으므로 상세 내용을 시각화할 수 없다. 따라서 추려진 집들이 어느 법정동, 자치구에 얼마나 많이 분포하는지를 알아본다.


**고객이 원하는 조건을 충족하는 집들중, 면적 넓은 20개의 집이 어느 법정동, 자치구에 많이 분포하는지 알아본다. **

```python
sql_cond = (((df1['건물면적'] / 3.3) >= 50) & (df1['건물주용도'] == '단독주택') &
            (df1['물건금액'] <= 4500000000) & (df1['물건금액'] >= 4000000000))
cus = df1[sql_cond]
cus_3_gu = (cus.groupby('자치구명').size().reset_index(name='적합 건수')
            .sort_values('적합 건수', ascending=False).reset_index(drop=True))
cus_3_gu
```

```python
cus_3 = (cus.groupby(['법정동명']).agg(자치구명=('자치구명', 'first'), 적합건수=('ID', 'size'))
         .reset_index().rename(columns={'적합건수': '적합 건수'}))
cus_3 = cus_3[['자치구명', '법정동명', '적합 건수']].sort_values('적합 건수', ascending=False).reset_index(drop=True)
cus_3
```

```python
cus_3_20 = cus_3.head(20)
cus_3_20
```

```python
# 법정동별 (자치구 중심 기반) 위경도. 원본은 구글 지도 API로 동을 지오코딩했다.
lat_lng_list = []
for _, row in cus_3.iterrows():
    la, lo = gu_geocode(row['법정동명'], gu=row['자치구명'])
    lat_lng_list.append((row['법정동명'], la, lo))
lat_lng_list = pd.DataFrame(lat_lng_list, columns=['법정동명', '위도', '경도'])
lat_lng_list.head()
```

```python
LOCATION_SEOUL = [37.56, 127.00]

m = folium.Map(location=LOCATION_SEOUL, zoom_start=11)

for i in range(20):
    j = lat_lng_list['법정동명'][i]
    k = int(cus_3_20[cus_3_20['법정동명'] == j]['적합 건수'].iloc[0])
    folium.Marker([lat_lng_list['위도'][i], lat_lng_list['경도'][i]],
                  popup=k, tooltip=j,
                  icon=folium.Icon(color='red' if k >= 10 else 'orange' if k > 5 else 'gray',
                                   icon_color='white', icon='glyphicon glyphicon-home',
                                   prefix='glyphicon')).add_to(m)

folium.Choropleth(geo_data=municipalities_geo, data=cus_3_gu, fill_opacity=0.8,
                  columns=['자치구명', '적합 건수'], fill_color='Reds',
                  key_on='feature.properties.name').add_to(m)
m
```

# 4️⃣ 결론


서론에서 우리는 *"대체 어디서, 얼마나 많은 부동산이, 어떤 가격에 거래되고 있는가"* 라는 궁금증에서 출발했다. 서울특별시의 2018~2020년 실거래가·전월세 데이터를 직접 들여다본 결과는 다음과 같이 정리할 수 있다.

먼저 **거래량** 측면에서, 2018년에서 2019년 사이에는 거래유형과 관계없이 거래가 대체로 늘었지만, 2020년에 들어서면서 양상이 뚜렷하게 갈렸다. **매매 거래량은 오히려 크게 증가한 반면, 전세·준전세·월세·준월세 등 임대 거래량은 일제히 급감했다.** 이는 2020년 주택임대차보호법 개정과 저금리 기조 속에서 전세 매물을 구하기 어려웠던 '전세대란'의 상황, 그리고 '지금이 아니면 못 산다'는 매수 심리와 맞닿아 있는 것으로 보인다. 건물주용도별로 보면 월세·준월세는 단독/다가구의 비중이 압도적이었고, 보증금 규모가 큰 전세·준전세와 매매는 아파트의 비중이 가장 높았다.

**가격** 측면에서는, 매매가는 2018년에서 2020년으로 갈수록 더 높은 가격대에 분포가 두터워졌지만, 임대 유형의 거래가격에서는 뚜렷한 상승 추세를 확인하기는 어려웠다. 자치구 단위로 보면 매매·임대를 막론하고 **서초구·강남구·송파구(이른바 강남 3구)의 거래액 평균이 3년 내내 다른 자치구를 압도**했으며, 전반적으로 거래액 평균은 상승하는 흐름을 보였다.

마지막으로 **「찾아줘! 우리집」** 에서는 이렇게 파악한 데이터를 실제 의사결정에 활용해 보았다. 사회초년생(고객1), 신혼부부(고객2), 로또 당첨자(고객3)라는 서로 다른 상황의 고객이 제시한 조건(평수·건축연도·지역·건물유형·예산)을 **필터링 → 정렬 → 중복 제거**로 좁혀 추천 매물을 추리고, 이를 지도 위에 시각화했다. 단순한 거래 현황 파악을 넘어, 부동산 데이터가 *'나에게 맞는 집 찾기'* 라는 구체적인 문제에 어떻게 쓰일 수 있는지를 확인할 수 있었다.

정리하면, 최근 3년간 서울의 부동산은 2020년을 기점으로 **매매는 늘고 임대는 줄어드는** 거래량 변화와 함께, **강남 3구를 중심으로 한 높은 거래가격**이라는 특징을 보였다. 데이터를 통해 거래의 추이와 분포를 직접 확인하고, 조건에 맞는 집을 찾는 작업까지 수행하면서 부동산 데이터의 활용 가능성을 체감할 수 있었다.
